# Stage 0: workspace and reproducibility checks
Run this in your connected Colab kernel. This creates a tiny temporary probe under `/content/holycow-ml-outputs` and reports whether a mounted Drive or repository is visible. No downloads or training are started.


```python
import os, random, tempfile
from pathlib import Path
SEED = 42
random.seed(SEED)
try:
    import numpy as np
    np.random.seed(SEED)
except ImportError:
    print("NumPy missing")
try:
    import torch
    torch.manual_seed(SEED)
except ImportError:
    print("PyTorch missing")
print("Random seed:", SEED)
print("Colab working directory:", Path.cwd())
for p in [Path("/content/holycow"), Path("/content/drive/MyDrive"), Path("/Users/sam/Documents/GitHub/holycow")]:
    print(f"{p} exists: {p.exists()}")
out = Path("/content/holycow-ml-outputs")
out.mkdir(parents=True, exist_ok=True)
with tempfile.NamedTemporaryFile(mode="w+", dir=out, delete=True) as f:
    f.write("ok"); f.flush(); f.seek(0); assert f.read() == "ok"
print("Output directory read/write: OK:", out)
print("NOTE: /content is ephemeral; select persistent storage before experiments.")
```
